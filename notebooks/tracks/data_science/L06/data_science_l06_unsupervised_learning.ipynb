## 2. The core idea slowly

This is the **Uncertainty** stage. Your job is to track variation, lineage, randomness, retries, or sensitivity.

Uncertainty is not failure; hidden uncertainty is the failure.

Every level in this repository follows the same reasoning chain:

1. Observation: what is directly visible?
2. Representation: how is it encoded?
3. Operation: what do we compute or transform?
4. Evidence: what output supports the claim?
5. Limitation: what could still be wrong?


## 3. Mathematical spine

\[P(A\mid B) = \frac{P(B\mid A)P(A)}{P(B)}, \qquad \hat{y} = f_\theta(x)\]

Bayes' rule updates belief when evidence arrives. A prediction function maps inputs x to an estimated output, but only under assumptions learned from data.
