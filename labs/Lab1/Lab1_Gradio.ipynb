## Modificación de imagen con interfaz de Gradio

```python
import numpy as np

def sepia(image):
    if image is None:
        return None

    sepia_filter = np.array([
        [0.393, 0.769, 0.189],
        [0.349, 0.686, 0.168],
        [0.272, 0.534, 0.131]
    ])

    image = image.astype(np.float32)
    sepia_img = image.dot(sepia_filter.T)
    return np.clip(sepia_img, 0, 255).astype(np.uint8)

```

```python
gr.close_all()

demo = gr.Interface(
    fn=sepia, 
    inputs=gr.Image(sources=["upload", "webcam", "clipboard"]), # Habilita los tres botones
    outputs="image"
)

demo.launch()
```

## Parámetros de la interfaz
 * **title**: agrega el título en la parte superior de la interfaz.
 * **description**: agrega una descripción debajo del título; se puede utilizar Markdown y HTML.
 * **examples**: permite entregar ejemplos de entrada
 * **inputs**: puede ser una lista de tipos de entrada, los cuales se mapean a la función de predicción.
 * **outputs**: puede ser una lista de tipos de salida, los cuales se mapean desde la salida de la función de predicción.
 
 
 Gradio permite definir diferentes tipos de elementos para agregar en la interfaz web, en este caso se utiliza un *radiobutton* para permitir al usuario seleccionar entre una gama de opciones.

```python
import gradio as gr

def calculator(num1, operation, num2):
    if operation == "add":
        return num1 + num2
    elif operation == "subtract":
        return num1 - num2
    elif operation == "multiply":
        return num1 * num2
    elif operation == "divide":
        if num2 == 0:
            raise gr.Error("No se puede dividir entre cero.")
        return num1 / num2
    return None

gr.close_all()
demo = gr.Interface(
    fn=calculator,
    inputs=[
        gr.Number(value=4),
        gr.Radio(["add", "subtract", "multiply", "divide"]),
        "number",
    ],
    outputs="number",
    examples=[
        [5, "add", 3],
        [4, "divide", 2],
        [-4, "multiply", 2.5],
        [0, "subtract", 1.2],
    ],
    title="Test Calculator",
    description="Esto es una **calculadora de juguete**, creada con Gradio",
)

demo.launch()

```

## Bloques para la interfaz gráfica

Permite personalizar más el diseño de la interfaz.
Se necesitan los mismos 3 elementos básicos.
Consideraciones:  
1. Blocks permite generar aplicaciones web que combinan markdown, HTML, botones y componentes interactivos simplemente instanciando objetos en Python dentro del contexto de with. El orden en el que se instancian los componentes importa, ya que cada elemento es renderizado en la aplicación en el orden en el que fueron creados. 
2. Puedes definir funciones regulares de Python en cualquier parte de tu código y ejecutarlas usando Blocks. En el ejemplo, se define una función simple que retorna un mensaje de bienvenida, pero en la práctica es posible usar grandes modelos de ML. 
3. Es posible asignar eventos a los componentes de Blocks. Esto ejecutará tu función cuando el componente haya sido clickeado/cambiado/etc. Cuando se asigna un evento, se pasan tres parámetros: 
    - **fn**: la función a ejecutar.
    - inputs : lista de componentes de entrada
    - outputs : lista de componentes de salida 
 4. Blocks infiere de forma automática si un componente es interactivo (acepta input del usuario) o no, basado en los triggers definidos. En nuestro ejemplo, la primera entrada de texto no es interactiva, dado que el valor de salida no es usado como input. En algunos casos es deseable configurar esto, lo que es posible definiendo el booleano *interactive*, un parámetro que todo componente acepta. 
5. Es posible programar y desplegar Blocks en cualquier parte: jupyter notebooks, colab notebooks, o cualquier IDE de Python regular ya que Gradio usa el intérprete estándar de Python. También puedes compartir *Blocks* configurando `share=True` en `launch()`.




```python
import gradio as gr

def saludo(nombre):
    return f"bienvenido a Gradio Blocks, {nombre}!"


gr.close_all()

with gr.Blocks() as demo:
    gr.Markdown(
        """
    # Hola Mundo!
    Escribe algo debajo para el output.
    """
    )

    inp = gr.Textbox(placeholder="¿Cuál es tu nombre?")
    out = gr.Textbox()

    inp.change(
        fn=saludo,
        inputs=inp,
        outputs=out,
    )

demo.launch()


```

## Uso de pestañas (*Tabs*)

**¿Qué son las pestañas en Gradio?**

Las pestañas en Gradio son una forma de organizar diferentes interfaces de usuario dentro de una sola aplicación web. Permiten dividir una aplicación en secciones lógicas, lo que facilita la navegación y el uso.

**¿Cómo usar pestañas en Gradio?**

Para usar pestañas en Gradio, necesitas el componente `gr.Tabs()` y, dentro, cada sección con `gr.Tab("Nombre")` (en versiones antiguas aparecía como `TabItem`; la API actual usa `Tab`).

```python
import numpy as np
import gradio as gr

demo = gr.Blocks()

def flip_text(x):
    return x[::-1]

def flip_image(image):
    return np.fliplr(image)

gr.close_all()

with demo:
    gr.Markdown("Flip *text* or *image* files using Python")
    with gr.Tabs():
        with gr.Tab("Flip Text"):
            text_input = gr.Textbox()
            text_output = gr.Textbox()
            text_button = gr.Button("Flip")
        with gr.Tab("Flip Image"):
            with gr.Row():
                image_input = gr.Image()
                image_output = gr.Image()
            image_button = gr.Button("Flip")
            
    text_button.click(flip_text, inputs=text_input, outputs=text_output)
    image_button.click(flip_image, inputs=image_input, outputs=image_output)
    
demo.launch()
```

```python

import gradio as gr
from math import sqrt
import matplotlib.pyplot as plt
import numpy as np
import plotly.express as px
import pandas as pd
import altair

def outbreak(plot_type, r, month, countries, social_distancing):
    months = ["January", "February", "March", "April", "May"]
    m = months.index(month)
    start_day = 30 * m
    final_day = 30 * (m + 1)
    x = np.arange(start_day, final_day + 1)
    pop_count = {"USA": 350, "Canada": 40, "Mexico": 300, "UK": 120}
    if social_distancing:
        r = sqrt(r)
    df = pd.DataFrame({"day": x})
    for country in countries:
        df[country] = x ** (r) * (pop_count[country] + 1)

    if plot_type == "Matplotlib":
        fig = plt.figure()
        plt.plot(df["day"], df[countries].to_numpy())
        plt.title("Outbreak in " + month)
        plt.ylabel("Cases")
        plt.xlabel("Days since Day 0")
        plt.legend(countries)
        return fig
    elif plot_type == "Plotly":
        fig = px.line(df, x="day", y=countries)
        fig.update_layout(
            title="Outbreak in " + month,
            xaxis_title="Days since Day 0",
            yaxis_title="Cases",
        )
        return fig
    elif plot_type == "Altair":
        df = df.melt(id_vars="day").rename(columns={"variable": "country"})
        fig = altair.Chart(df).mark_line().encode(x="day", y='value', color='country')
        return fig
    else:
        raise ValueError("A plot type must be selected")

gr.close_all()

inputs = [
    gr.Dropdown(["Matplotlib", "Plotly", "Altair"], label="Plot Type"),
    gr.Slider(1, 4, 3.2, label="R"),
    gr.Dropdown(["January", "February", "March", "April", "May"], label="Month"),
    gr.CheckboxGroup(
        ["USA", "Canada", "Mexico", "UK"], label="Countries", value=["USA", "Canada"]
    ),
    gr.Checkbox(label="Social Distancing?"),
]
outputs = gr.Plot()

demo = gr.Interface(
    fn=outbreak,
    inputs=inputs,
    outputs=outputs,
    examples=[
        ["Matplotlib", 2, "March", ["Mexico", "UK"], True],
        ["Altair", 2, "March", ["Mexico", "Canada"], True],
        ["Plotly", 3.6, "February", ["Canada", "Mexico", "UK"], False],
    ],
    cache_examples=True,
)

if __name__ == "__main__":
    demo.launch()

```

```python
import inspect
import gradio as gr


def _make_blocks(title, theme):
    """Gradio 6+: el tema va en launch(); en versiones anteriores, en Blocks()."""
    kw = {"title": title}
    if "theme" not in inspect.signature(gr.Blocks.launch).parameters:
        kw["theme"] = theme
    return kw


def _launch(demo, theme=None, css=None):
    """Pasa theme/css a launch() solo si la firma de tu versión lo admite.
    Devuelve el resultado de launch() para que Jupyter incruste la interfaz."""
    kw = {}
    sig = inspect.signature(demo.launch)
    if theme is not None and "theme" in sig.parameters:
        kw["theme"] = theme
    if css is not None and str(css).strip() and "css" in sig.parameters:
        kw["css"] = css
    return demo.launch(**kw)
```

### Problema 1

Realice una aplicación con Gradio que reciba 3 elementos numéricos como entrada y produzca 3 elementos de salida.

- El primer elemento debe ser la suma de los dos primeros números elevada al tercer número: $(A+B)^C$.
- Segundo elemento: a partir de la suma de los dos primeros, restar el cuadrado de la diferencia de los dos últimos números.
- Tercer elemento: la división del **mayor** de los números al cubo entre el **valor intermedio** (al ordenar los tres números).

> **Nota (implementación):** Los tres valores del enunciado se muestran en tres cuadros de texto. El bloque **JSON** del cuaderno es una vista estructurada adicional del mismo resultado, no una cuarta magnitud del problema.

```python
import gradio as gr

LAB_CSS = """
.gradio-container { max-width: 960px !important; margin: auto !important; }
"""


def problema1_calculos(a, b, c):
    if a is None or b is None or c is None:
        raise gr.Error("Introduce los tres números (A, B y C).")
    salida1 = (a + b) ** c
    salida2 = (a + b) - (b - c) ** 2
    vals = sorted([a, b, c])
    mid, mx = vals[1], vals[2]
    if mid == 0:
        raise gr.Error("El número intermedio es 0; no se puede dividir.")
    salida3 = (mx**3) / mid
    return salida1, salida2, salida3


def run_problema1(a, b, c):
    s1, s2, s3 = problema1_calculos(a, b, c)
    resumen = {
        "(A+B)^C": float(s1),
        "(A+B)−(B−C)²": float(s2),
        "(mayor)³ ÷ intermedio": float(s3),
    }
    return f"{s1:,.12g}", f"{s2:,.12g}", f"{s3:,.12g}", resumen


def limpiar_p1():
    return (
        gr.update(value=1.0),
        gr.update(value=2.0),
        gr.update(value=3.0),
        "",
        "",
        "",
        gr.update(value={}),
    )


gr.close_all()
THEME = gr.themes.Soft(primary_hue="amber", neutral_hue="stone")

with gr.Blocks(**_make_blocks("Problema 1 · Calculadora triple", THEME)) as demo:
    gr.Markdown(
        "## Laboratorio 1 · Problema 1\n"
        "Panel tipo **dashboard**: entradas a la izquierda, resultados a la derecha. "
        "Usa **Ejemplos** o **Calcular**."
    )
    with gr.Accordion("Fórmulas del enunciado", open=False):
        gr.Markdown(
            "- **Salida 1:** $(A+B)^C$\n"
            "- **Salida 2:** $(A+B) - (B-C)^2$\n"
            "- **Salida 3:** máximo al cubo ÷ valor **intermedio** al ordenar A, B, C."
        )
    with gr.Row():
        with gr.Column(scale=1):
            with gr.Group():
                gr.Markdown("### Entradas")
                a_in = gr.Number(label="A", value=1, precision=4, info="Primer número")
                b_in = gr.Number(label="B", value=2, precision=4, info="Segundo número")
                c_in = gr.Number(label="C", value=3, precision=4, info="Tercer número")
                with gr.Row():
                    calc_btn = gr.Button("Calcular", variant="primary", scale=2)
                    limp_btn = gr.Button("Limpiar", variant="secondary", scale=1)
        with gr.Column(scale=1):
            with gr.Group():
                gr.Markdown("### Resultados")
                out1 = gr.Textbox(label="(A+B)^C", interactive=False, lines=1)
                out2 = gr.Textbox(label="(A+B) − (B−C)²", interactive=False, lines=1)
                out3 = gr.Textbox(label="(mayor)³ ÷ intermedio", interactive=False, lines=1)
                out_json = gr.JSON(label="Vista estructurada")
    gr.Examples(
        examples=[[1, 2, 3], [0, 5, 2], [2.5, -1, 4]],
        inputs=[a_in, b_in, c_in],
        outputs=[out1, out2, out3, out_json],
        fn=run_problema1,
        label="Probar con un ejemplo",
        run_on_click=True,
    )
    outs = [out1, out2, out3, out_json]
    calc_btn.click(run_problema1, [a_in, b_in, c_in], outs)
    limp_btn.click(limpiar_p1, None, [a_in, b_in, c_in] + outs)

import os
if os.environ.get("SKIP_LAUNCH", "0") != "1":
    _launch(demo, THEME, LAB_CSS)
```

### Problema 2

Defina una aplicación con Gradio que reciba 2 listas de entre 5 y 8 números.
Utilizando **Matplotlib**, grafique las listas en formato X vs. Y y muestre el resultado en el panel correspondiente.

```python
import gradio as gr
import matplotlib.pyplot as plt
import numpy as np

LAB_CSS = """
.gradio-container { max-width: 960px !important; margin: auto !important; }
"""


def _parse_list(text):
    parts = [p.strip() for p in text.replace(";", ",").split(",") if p.strip()]
    return [float(p) for p in parts]


def problema2_grafico(txt_x, txt_y, estilo, paleta_nombre):
    try:
        x = _parse_list(txt_x)
        y = _parse_list(txt_y)
    except ValueError:
        raise gr.Error("Usa solo números separados por comas (ej.: 1, 2, 3, 4, 5).")
    if len(x) != len(y):
        raise gr.Error("Las dos listas deben tener la misma longitud.")
    n = len(x)
    if not (5 <= n <= 8):
        raise gr.Error(f"Cada lista debe tener entre 5 y 8 números (tienes {n}).")

    palettes = {
        "Océano": ("#0284c7", "#facc15", "#0369a1"),
        "Atardecer": ("#e11d48", "#fb7185", "#9f1239"),
        "Bosque": ("#16a34a", "#fde047", "#14532d"),
    }
    col, mfc, mec = palettes.get(paleta_nombre, palettes["Océano"])

    fig, ax = plt.subplots(figsize=(8.4, 5), dpi=115)
    ax.set_facecolor("#f8fafc")
    fig.patch.set_facecolor("#ffffff")
    if estilo == "Solo puntos":
        ax.scatter(x, y, s=120, c=col, edgecolors=mec, linewidths=1.2, zorder=3, label="Y vs. X")
    elif estilo == "Línea y puntos":
        ax.plot(x, y, linestyle="-", linewidth=2.2, color=col, alpha=0.85, label="Línea")
        ax.scatter(x, y, s=95, facecolors=mfc, edgecolors=mec, linewidths=1.1, zorder=3, label="Puntos")
    else:
        ax.plot(
            x, y, marker="o", linewidth=2.6, markersize=9,
            color=col, markerfacecolor=mfc, markeredgecolor=mec, markeredgewidth=1.3,
            label="Serie Y vs. X",
        )
    ax.set_xlabel("Lista X", fontsize=11)
    ax.set_ylabel("Lista Y", fontsize=11)
    ax.set_title("Problema 2 · Visualización X vs. Y", fontsize=13, fontweight="bold")
    ax.grid(True, alpha=0.35, linestyle="--")
    ax.legend(framealpha=0.92, loc="best")
    fig.tight_layout()
    return fig


def _ejemplo_lists(vx, vy):
    return problema2_grafico(vx, vy, "Línea", "Océano")


gr.close_all()
THEME = gr.themes.Ocean(primary_hue="cyan")

with gr.Blocks(**_make_blocks("Problema 2 · Visualizador X–Y", THEME)) as demo:
    gr.Markdown(
        "## Laboratorio 1 · Problema 2\n"
        "Elige **tipo de trazo** y **paleta**, introduce dos listas (5–8 números) y genera el gráfico."
    )
    with gr.Group():
        gr.Markdown("### Datos y estilo")
        estilo_plot = gr.Radio(
            ["Línea", "Solo puntos", "Línea y puntos"],
            value="Línea",
            label="Tipo de gráfico",
        )
        paleta = gr.Dropdown(
            choices=["Océano", "Atardecer", "Bosque"],
            value="Océano",
            label="Paleta de colores",
        )
        tx = gr.Textbox(
            label="Lista X",
            placeholder="ej.: 1, 2, 3, 4, 5, 6",
            value="1, 2, 3, 4, 5, 6",
            info="Entre 5 y 8 números separados por coma.",
        )
        ty = gr.Textbox(
            label="Lista Y",
            placeholder="ej.: 2, 4, 1, 5, 3, 0",
            value="2, 4, 1, 5, 3, 0",
            info="Misma longitud que X.",
        )
        go = gr.Button("Generar gráfico", variant="primary")
    with gr.Group():
        gr.Markdown("### Gráfico")
        plot_out = gr.Plot(label="Matplotlib")
    gr.Examples(
        examples=[
            ["1, 2, 3, 4, 5", "2, 4, 1, 5, 3"],
            ["0.5, 1, 1.5, 2, 2.5, 3", "1, 1, 2, 3, 5, 8"],
        ],
        inputs=[tx, ty],
        outputs=[plot_out],
        fn=_ejemplo_lists,
        label="Ejemplos de listas",
        run_on_click=True,
    )
    go.click(problema2_grafico, [tx, ty, estilo_plot, paleta], plot_out)

import os
if os.environ.get("SKIP_LAUNCH", "0") != "1":
    _launch(demo, THEME, LAB_CSS)
```

La aplicación debe aplicar alguna **operación** a la entrada y reflejarla en la salida. Puede partir del chatbot o de imagen siempre que **cambie** el comportamiento respecto al ejemplo del laboratorio.

```python
import numpy as np
import gradio as gr

LAB_CSS = """
.gradio-container { max-width: 960px !important; margin: auto !important; }
"""


def problema3_audio_normalizado(audio, target_peak):
    if audio is None:
        return None, "**Sube o graba un audio.**"
    sr, data = audio
    if data is None or data.size == 0:
        return None, "**Audio vacío.**"
    x = np.asarray(data, dtype=np.float32)
    if x.ndim > 1:
        x = x.mean(axis=1)
    peak_before = float(np.max(np.abs(x)))
    if peak_before == 0:
        return (sr, x), "**Silencio (pico = 0).**"
    t = float(target_peak)
    y = (x / peak_before * t).astype(np.float32)
    peak_after = float(np.max(np.abs(y)))
    dur = len(x) / float(sr)
    stats = (
        f"**Duración:** {dur:.2f} s\n"
        f"**Muestras:** {len(x)} · **Frecuencia:** {int(sr)} Hz\n"
        f"**Pico antes:** {peak_before:.5f}\n"
        f"**Pico después:** {peak_after:.5f} (objetivo **{t}**)\n"
        f"**Procesado:** canal mono (promedio si era estéreo)"
    )
    return (sr, y), stats


gr.close_all()
THEME = gr.themes.Citrus(primary_hue="lime")

with gr.Blocks(**_make_blocks("Problema 3 · Audio normalizado", THEME)) as demo:
    gr.Markdown(
        "## Laboratorio 1 · Problema 3\n"
        "**Normalizador de onda** con panel de estadísticas y pico objetivo ajustable."
    )
    with gr.Accordion("Qué hace la normalización", open=False):
        gr.Markdown(
            "Divide la señal por su pico actual y escala al **objetivo** "
            "(evita clipping al reproducir). Si hay varios canales, se promedian a mono."
        )
    target_peak = gr.Slider(0.3, 1.0, value=0.95, step=0.05, label="Pico objetivo (amplitud máx.)")
    with gr.Row():
        with gr.Column():
            with gr.Group():
                gr.Markdown("### Entrada")
                au_in = gr.Audio(
                    label="Audio",
                    type="numpy",
                    sources=["upload", "microphone"],
                )
                norm_btn = gr.Button("Normalizar", variant="primary")
        with gr.Column():
            with gr.Group():
                gr.Markdown("### Salida")
                au_out = gr.Audio(label="Audio normalizado", type="numpy")
                stats_md = gr.Markdown()
    norm_btn.click(problema3_audio_normalizado, [au_in, target_peak], [au_out, stats_md])

import os
if os.environ.get("SKIP_LAUNCH", "0") != "1":
    _launch(demo, THEME, LAB_CSS)
```

```python
import json
import gradio as gr

LAB_CSS = """
.gradio-container { max-width: 1080px !important; margin: auto !important; }
.p4-hero {
  background: linear-gradient(125deg, #5b21b6 0%, #7c3aed 40%, #0891b2 100%) !important;
  border-radius: 16px !important;
  padding: 1.25rem 1.5rem !important;
  margin-bottom: 0.9rem !important;
  box-shadow: 0 12px 40px rgba(91, 33, 182, 0.28) !important;
}
.p4-hero, .p4-hero h2, .p4-hero p, .p4-hero em, .p4-hero strong { color: #f8fafc !important; }
.p4-hero h2 { margin: 0.2em 0 0.35em 0 !important; font-weight: 700 !important; letter-spacing: -0.02em !important; }
.p4-badge {
  display: inline-block !important;
  background: rgba(255,255,255,0.22) !important;
  color: #fff !important;
  padding: 0.25rem 0.65rem !important;
  border-radius: 999px !important;
  font-size: 0.72rem !important;
  font-weight: 700 !important;
  letter-spacing: 0.08em !important;
  text-transform: uppercase !important;
}
.p4-panel-in {
  border: 1px solid rgba(124, 58, 237, 0.35) !important;
  border-radius: 14px !important;
  background: linear-gradient(180deg, #faf5ff 0%, #f5f3ff 55%, #ffffff 100%) !important;
  padding: 0.85rem !important;
  box-shadow: 0 4px 18px rgba(124, 58, 237, 0.08) !important;
}
.p4-panel-out {
  border: 1px solid rgba(8, 145, 178, 0.45) !important;
  border-radius: 14px !important;
  background: linear-gradient(180deg, #ecfeff 0%, #f0fdfa 100%) !important;
  padding: 0.85rem !important;
  box-shadow: 0 4px 18px rgba(8, 145, 178, 0.12) !important;
}
.p4-tabs-wrap [role="tablist"] button[aria-selected="true"] {
  color: #7c3aed !important;
  border-bottom-color: #7c3aed !important;
}
"""


def problema4_format(blob, texto_pegar, modo):
    raw = None
    origen = ""
    if blob is not None:
        if isinstance(blob, str):
            path = blob
        elif hasattr(blob, "name"):
            path = blob.name
        else:
            path = str(blob)
        try:
            with open(path, "r", encoding="utf-8") as f:
                raw = f.read()
            origen = "(desde archivo)"
        except OSError as e:
            err = {"error": "archivo", "detalle": str(e)}
            msg = f"**Error de archivo:** {e}"
            return json.dumps(err, ensure_ascii=False, indent=2), err, msg
    if raw is None and texto_pegar and texto_pegar.strip():
        raw = texto_pegar.strip()
        origen = "(desde texto pegado)"
    if not raw:
        msg = "**Sube un `.json` o pega JSON en el cuadro de texto y pulsa _Analizar JSON_.**"
        return "", None, msg
    try:
        data = json.loads(raw)
    except json.JSONDecodeError as e:
        err = {"error": "JSON inválido", "detalle": str(e)}
        msg = f"**JSON inválido:** {e}"
        return json.dumps(err, ensure_ascii=False, indent=2), err, msg
    if modo == "Minificar":
        pretty = json.dumps(data, ensure_ascii=False, separators=(",", ":"))
    else:
        pretty = json.dumps(data, ensure_ascii=False, indent=2)
    if isinstance(data, dict):
        tip_base = f"**Tipo:** objeto · **Claves:** {len(data)}"
    elif isinstance(data, list):
        tip_base = f"**Tipo:** lista · **Elementos:** {len(data)}"
    else:
        tip_base = f"**Tipo:** {type(data).__name__}"
    msg = f"{tip_base} {origen}".strip()
    return pretty, data, msg


SAMPLE_JSON = '{"nombre": "Demo", "nota": 7.5, "tags": ["a", "b"]}'


gr.close_all()
THEME = gr.themes.Soft(
    primary_hue="violet",
    secondary_hue="cyan",
    neutral_hue="slate",
    font=[gr.themes.GoogleFont("DM Sans"), "system-ui", "sans-serif"],
    font_mono=[gr.themes.GoogleFont("JetBrains Mono"), "ui-monospace", "monospace"],
)

with gr.Blocks(**_make_blocks("Problema 4 · JSON Studio", THEME)) as demo:
    with gr.Column(elem_classes=["p4-hero"]):
        gr.Markdown(
            '<span class="p4-badge">Laboratorio 1</span>\n\n'
            "## JSON Studio · Problema 4\n"
            "Formatea, **minifica** y explora tu JSON con vista de _código_ y _árbol_."
        )
    with gr.Row():
        with gr.Column(scale=7):
            with gr.Group(elem_classes=["p4-panel-in"]):
                gr.Markdown("### Entrada — archivo o texto")
                modo = gr.Radio(
                    ["Pretty", "Minificar"],
                    value="Pretty",
                    label="Modo de salida",
                    info="Pretty = indentado legible · Minificar = una sola línea.",
                )
                f_in = gr.File(
                    label="Archivo .json",
                    file_types=[".json"],
                    type="filepath",
                    show_label=True,
                )
                pegar = gr.Textbox(
                    label="O pega JSON aquí",
                    placeholder='{"clave": "valor"}',
                    lines=8,
                    autofocus=True,
                )
                with gr.Row():
                    fmt_btn = gr.Button("Analizar JSON", variant="primary", scale=2)
                    ej_btn = gr.Button("Cargar ejemplo", variant="secondary", scale=1)
        with gr.Column(scale=5):
            with gr.Group(elem_classes=["p4-panel-out"]):
                gr.Markdown("### Resultado")
                hint = gr.Markdown("Lista para analizar tu JSON.")
                with gr.Column(elem_classes=["p4-tabs-wrap"]):
                    with gr.Tabs():
                        with gr.Tab("Código"):
                            code_out = gr.Code(language="json", lines=20, interactive=False)
                        with gr.Tab("Árbol"):
                            tree_out = gr.JSON(label="Vista de objeto")

    def load_sample():
        return gr.update(value=SAMPLE_JSON)

    fmt_btn.click(problema4_format, [f_in, pegar, modo], [code_out, tree_out, hint])
    ej_btn.click(load_sample, None, pegar)

import os
if os.environ.get("SKIP_LAUNCH", "0") != "1":
    _launch(demo, THEME, LAB_CSS)
```
