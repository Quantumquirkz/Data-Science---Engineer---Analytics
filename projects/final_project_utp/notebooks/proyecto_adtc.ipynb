# 1. Preparación del entorno y carga del conjunto de datos

```python
# Cargar librerías necesarias

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import missingno as msno
from upsetplot import UpSet, from_indicators
from sklearn.preprocessing import LabelEncoder
from pathlib import Path
pd.set_option('display.max_columns', None)
sns.set_theme(style='whitegrid', context='notebook', palette='Set2')
plt.rcParams.update({
    'figure.dpi': 180,
    'savefig.dpi': 300,
    'savefig.bbox': 'tight',
    'font.size': 11,
    'axes.titlesize': 13,
    'axes.labelsize': 11,
    'xtick.labelsize': 10,
    'ytick.labelsize': 10,
})

def find_project_root() -> Path:
    """Return the repository root regardless of the current working directory."""
    cwd = Path.cwd().resolve()
    for anchor in [cwd, *cwd.parents]:
        direct = anchor / "data" / "raw" / "student-mat.csv"
        if direct.exists():
            return anchor
        nested = anchor / "projects" / "final_project_utp" / "data" / "raw" / "student-mat.csv"
        if nested.exists():
            return anchor / "projects" / "final_project_utp"
    raise FileNotFoundError("No se encontró la raíz del proyecto con la carpeta data/raw.")

PROJECT_ROOT = find_project_root()
```

# 4. Transformación de variables categóricas para modelado

```python
# Para modelos de ML es necesario convertir los datos categóricos a numéricos
df.select_dtypes(include='object').columns
```

```python
# Se convertirán las columnas de valores de texto a numéricas
le = LabelEncoder()

columnas_categoricas = df.select_dtypes(
    include=['object', 'string']
).columns

for columna in columnas_categoricas:
    df[columna] = le.fit_transform(df[columna])

df.head() 
```

```python
# Verificación del tipo de los datos
df.info()
```

Como podemos analizar, todos los datos de tipo de texto han sido convertidas en datos de tipo numéricos para ser utilizadas más adelante en modelos de ML
