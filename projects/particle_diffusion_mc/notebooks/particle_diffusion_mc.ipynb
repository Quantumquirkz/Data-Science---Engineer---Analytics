# Particle diffusion Monte Carlo

We approximate Brownian motion with exact Gaussian increments and compare the empirical law of `X_T` to the **fundamental solution** of the heat equation

`u_t = (sigma^2 / 2) u_xx` with `u(0,x) = delta(x)`.

That solution is the Gaussian density with variance `sigma^2 T`, the same as the law of `X_T` for `dX = sigma dW` (Ito).

**Stochastic calculus intuition:** white noise integrated in time produces independent increments; scaling `sqrt(dt)` keeps finite quadratic variation. **Monte Carlo:** many paths estimate expectations and entire distributions. **MSD:** `E[X_t^2] = sigma^2 t` in 1D grows linearly, the signature of pure diffusion.
